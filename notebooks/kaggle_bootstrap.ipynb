# tuned - Kaggle bootstrap

This notebook pulls the latest code from [Anant-T/Tuned](https://github.com/Anant-T/Tuned) on every run.

**Before the first run:** turn on `Settings -> Internet` in the notebook sidebar, or the clone will fail.

Workflow: edit locally -> `git push` -> re-run cell 1 here.

```python
# Cell 1 - the bridge. Safe to re-run; pulls the newest commit each time.
import os, sys, subprocess

REPO = "https://github.com/Anant-T/Tuned"
DIR = "/kaggle/temp/Tuned"  # scratch, NOT /kaggle/working - keeps source out of your notebook output

if os.path.isdir(f"{DIR}/.git"):
    subprocess.run(["git", "-C", DIR, "pull", "--ff-only"], check=True)
else:
    os.makedirs(os.path.dirname(DIR), exist_ok=True)
    subprocess.run(["git", "clone", "--depth", "1", REPO, DIR], check=True)

src = f"{DIR}/src"
if src not in sys.path:
    sys.path.insert(0, src)

print(f"ready: {src}")
```

```python
# Cell 2 - proves the bridge works.
# Bump __version__ locally, push, re-run cell 1, re-run this: the number should change.
import importlib
import tuned

importlib.reload(tuned)  # picks up a fresh pull without restarting the kernel
tuned.where_am_i()
```
